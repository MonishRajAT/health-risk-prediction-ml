```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data
df = pd.read_csv("liver.csv")

# Basic info
df.head()
df.info()
```

```python
# Geneder distribution 
sns.countplot(x = "gender", data = df)
plt.title("gender distribution")
plt.show()
```

```python
# Target distribution
sns.countplot(x = "is_patient", data = df)
plt.title("Liver Disease distribution")
plt.show()
```

```python
# Bilirubin distribution
sns.boxplot(x = "is_patient", y = "tot_bilirubin", data = df)
plt.title("Total Bilirubin vs Liver Disease")
plt.show()
```
